```python
from pathlib import Path

ref_dir = Path("dataset/dataset_list")
```

```python
train_datasets = ref_dir / "btcv_train.txt"
val_datasets = ref_dir / "btcv_val.txt"
test_datasets = ref_dir / "btcv_test.txt"
```

```python
train_set = []
with open(train_datasets, "r") as f:
    lines = f.readlines()
    for line in lines:
        line = line.strip("\n")
        img,label = line.split("\t")
        train_set.append((Path(img).name, Path(label).name))
        
val_set = []
with open(val_datasets, "r") as f:
    lines = f.readlines()
    for line in lines:
        line = line.strip("\n")
        img,label = line.split("\t")
        val_set.append((Path(img).name, Path(label).name))
        
test_set = []
with open(test_datasets, "r") as f:
    lines = f.readlines()
    for line in lines:
        line = line.strip("\n")
        img,label = line.split("\t")
        test_set.append((Path(img).name, Path(label).name))
```

```python
import shutil
from tqdm import tqdm

raw_dataset_dir = Path("data/RawData/Training")
new_dataset_dir = Path("data/01_Multi-Atlas_Labeling")
for img in tqdm(list((raw_dataset_dir / "img").iterdir())):
    img_name = img.name
    label_path = raw_dataset_dir / "label" / img_name.replace("img","label")
    
    shutil.copy(img,new_dataset_dir / "img" / img_name)
    shutil.copy(label_path,new_dataset_dir / "label" / label_path.name)
        
   
```

#### 重新生成 btcv_train.txt

```python
base_dir = Path("data")
```

```python
raw_dataset_dir = Path("data/RawData/Training")
new_dataset_dir = Path("data/01_Multi-Atlas_Labeling")

dataset_pair = []
for img in tqdm(list((raw_dataset_dir / "img").iterdir())):
    img_name = img.name
    label_path = raw_dataset_dir / "label" / img_name.replace("img","label")
    
    dataset_pair.append((base_dir / "01_Multi-Atlas_Labeling" / "img" / img_name,base_dir / "01_Multi-Atlas_Labeling" / "label" / label_path.name))
```

```python
dataset_pair[0]
```

```python
from sklearn.model_selection import train_test_split

train_pair, test_pair = train_test_split(dataset_pair, test_size=0.2, random_state=42)
```

```python
train_pair[0]
```

```python
train_file = Path("dataset/dataset_list/btcv_train_new.txt")
val_file = Path("dataset/dataset_list/btcv_val_new.txt")

with open(train_file, "w") as f:
    for img, label in train_pair:
        f.write(f"{str(img)}\t{str(label)}\n")
        
with open(val_file, "w") as f:
    for img, label in test_pair:
        f.write(f"{str(img)}\t{str(label)}\n")
```
